DICOM Image

```python
import SimpleITK as sitk
import matplotlib.pyplot as plt

image = sitk.ReadImage("/content/drive/MyDrive/cancer/19.dcm")
array = sitk.GetArrayFromImage(image)

plt.imshow(array[0], cmap='gray')  # DICOMs are often 2D
plt.title("DICOM Image")
plt.axis("off")
plt.show()
```

```python
import nibabel as nib
import matplotlib.pyplot as plt
import numpy as np

nii = nib.load("/content/drive/MyDrive/3D dataset/segmentations/segmentation-0.nii")
data = nii.get_fdata()

slice_idx = 58  # your desired slice number
slice_data = data[:, :, slice_idx]

# Normalize for better contrast
norm_slice = (slice_data - np.min(slice_data)) / (np.max(slice_data) - np.min(slice_data) + 1e-8)

plt.imshow(norm_slice, cmap='gray')
plt.title(f"NIfTI Image (Slice {slice_idx})")
plt.axis('off')
plt.show()
```
